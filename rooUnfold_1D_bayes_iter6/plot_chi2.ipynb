```python
# %% [markdown]
# Imports

# %% 
import ROOT
import numpy as np
import matplotlib.pyplot as plt


# %%
root_file_name = "unfold_out_bayes.root"  # adjust path if needed

f = ROOT.TFile.Open(root_file_name)
if not f or f.IsZombie():
    raise RuntimeError(f"Could not open {root_file_name}")

chi2_param = f.Get("chi2_vs_truth")
if not chi2_param:
    raise RuntimeError("TParameter<double> 'chi2_vs_truth' not found in file")

chi2_val = chi2_param.GetVal()
print("chi2_vs_truth =", chi2_val)
```

```python
# %%
n_iter = 1  # change this if your iteration number is different

plt.figure()
plt.scatter([n_iter], [chi2_val])
plt.xlabel("Iteration")
plt.ylabel(r"$\chi^2$ (unfolded vs MC truth)")
plt.title("Bayes unfolding: chi2_vs_truth")
plt.grid(True)
plt.show()
```
